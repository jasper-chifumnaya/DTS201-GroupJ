```python
df['Highest_State'] = df['Highest'].str.extract(r'^(.+?)\s*\(')
df['Highest_Price'] = df['Highest'].str.extract(r'\(([\d.]+)\)').astype(float)

df['Lowest_State'] = df['Lowest'].str.extract(r'^(.+?)\s*\(')
df['Lowest_Price'] = df['Lowest'].str.extract(r'\(([\d.]+)\)').astype(float)
```

Overall descriptive statistics

```python
overall_stats = df.groupby('Item Label')['Avg_Price'].agg(
    Mean='mean',
    Median='median',
    Minimum='min',
    Maximum='max',
    Std_Dev='std'
).reset_index()
```

```python
overall_stats['CV'] = (
    overall_stats['Std_Dev'] / overall_stats['Mean']
) * 100
```

```python
overall_stats
```

```python
Path('outputs').mkdir(exist_ok=True)

overall_stats.to_csv('../outputs/overall_summary_statistics.csv', index=False)
```

Yearly descriptive statistics

```python
yearly_stats = df.groupby(['Item Label', 'Year'])['Avg_Price'].agg(
    Mean='mean',
    Median='median',
    Minimum='min',
    Maximum='max',
    Std_Dev='std'
).reset_index()
```

```python
yearly_stats['CV'] = (
    yearly_stats['Std_Dev'] / yearly_stats['Mean']
) * 100
```

```python
yearly_stats
```

```python
yearly_stats.to_csv('../outputs/yearly_summary_statistics.csv', index=False)
```

Price surge analysis

```python
df_sorted = df.sort_values(['Item Label', 'Year', 'Month_Num'])
```

```python
surge = df_sorted.groupby('Item Label')['Avg_Price'].agg(
    First_Price='first',
    Last_Price='last'
).reset_index()
```

```python
surge['Price_Surge_%'] = (
    (surge['Last_Price'] - surge['First_Price'])
    / surge['First_Price']
) * 100
```

```python
top_5_surge = surge.sort_values(
    'Price_Surge_%', ascending=False
).head(5)
```

```python
top_5_surge
```

```python
top_5_surge.to_csv('../outputs/top_5_price_surge.csv', index=False)
```

MoM volatility analysis

```python
mom_volatility = df.groupby('Item Label')['MoM'].agg(
    Average_MoM='mean',
    MoM_Volatility='std'
).reset_index()
```

```python
top_5_volatility = mom_volatility.sort_values(
    'MoM_Volatility', ascending=False
).head(5)
```

```python
top_5_volatility
```

```python
top_5_volatility.to_csv(
    '../outputs/top_5_mom_volatility.csv', index=False
)
```

State frequency analysis

```python
highest_states = df.groupby('Item Label')['Highest_State'].agg(
    lambda x: x.value_counts().index[0]
).reset_index(name='Most_Frequent_Highest_State')
```

```python
lowest_states = df.groupby('Item Label')['Lowest_State'].agg(
    lambda x: x.value_counts().index[0]
).reset_index(name='Most_Frequent_Lowest_State')
```

```python
state_summary = pd.merge(
    highest_states,
    lowest_states,
    on='Item Label'
)
```

```python
state_summary
```

```python
state_summary.to_csv(
    '../outputs/state_frequency_summary.csv', index=False
)
```
